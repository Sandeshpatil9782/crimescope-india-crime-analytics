# District-wise IPC Crime Analysis in India (2001–2012)

## Notebook 2: Data Cleaning & Preprocessing


```python
import pandas as pd
import numpy as np

pd.set_option("display.max_columns", None)

df = pd.read_csv("crime_raw_backup1.csv")
```

```python
df.head()
```

```python
df.columns
```

```python
df.columns = (
    df.columns
    .str.strip()
    .str.upper()
    .str.replace(" ", "_")
    .str.replace("/", "_", regex=False)
    .str.replace("&", "AND", regex=False)
)
```

```python
df.columns.tolist()
```

```python
df.isnull().sum().sum()
```

```python
df.duplicated().sum()
```

```python
df[df["DISTRICT"].str.contains("TOTAL", case=False, na=False)]
```

```python
df = df[
    ~df["DISTRICT"].str.contains("TOTAL", case=False, na=False)
].copy()
```
